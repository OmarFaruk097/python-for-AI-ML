```python
inp = input()
numbers=inp.split()

x=int(numbers[0])
y=int(numbers[1])
z=int(numbers[2])

min=x
max=x
if y<min:
    min=y
if z<min:
    min=z
if y>max:
    max=y
if z>max:
    max=z
print("The maximum number is:", max)
print("The minimum number is:", min)

```

```python
n=int(input())

inp = input()
numbers=inp.split()
positive=0;negative=0; even=0;odd=0
for i in range(n):
    x=int(numbers[i])
    if x>0:
        positive+=1
    else:
        negative+=1
    if x%2==0:
        even+=1
    else:
        odd+=1
print("Positive numbers:", positive)
print("Negative numbers:", negative)
print("Even numbers:", even)
print("Odd numbers:", odd)
```
